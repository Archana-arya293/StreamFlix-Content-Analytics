```python
import pandas as pd
import os

print(os.listdir("phase3_clean_data"))
```

```python
#loading clean dataset


#path to phase3 cleaned dataset

phase3_data_path = "phase3_clean_data"

#loading clean datasets


subscribers = pd.read_csv(os.path.join(phase3_data_path,"subscribers_clean.csv"))

watch_history = pd.read_csv(os.path.join(phase3_data_path,"watch_history_clean.csv"))

titles = pd.read_csv(os.path.join(phase3_data_path,"titles_clean.csv"))

ratings = pd.read_csv(os.path.join(phase3_data_path,"ratings_clean.csv"))

reviews = pd.read_csv(os.path.join(phase3_data_path,"reviews_clean.csv"))

watchlist = pd.read_csv(os.path.join(phase3_data_path,"watchlist_clean.csv"))

print("All Cleaned Datasets Loaded Sucessfully.")

```

```python
#verifying the data

print("subscribers:", subscribers.shape)
print("watch_history:",watch_history.shape)
print("titles:",titles.shape)
print("ratings:",ratings.shape)
print("reviews:",reviews.shape)
print("watchlist:",watchlist.shape)
```

```python
# creating first KPI'S
#Total Watch Hours 
#For calculatinf formula is sum(watch_duration_min)/60


print(watch_history[["watch_duration_min"]].info())
```

```python
#Calculating Total Watch Hours

total_watch_hours = watch_history['watch_duration_min'].sum() / 60

print(f"Total Watch Hours: {total_watch_hours:,.2f}")
```

```python
#calculating 2nd KPI'S for that we'll use Total Subscribersx100
#Active Rate

print(subscribers.columns.tolist())


```

```python
#we have field is_active ,directly we can calculate Active Rate from the subscribers table

#validating


print(subscribers['is_active'].value_counts(dropna=False))
```

```python
# calculating Active Rate

total_subscribers = subscribers['subscriber_id'].nunique()
active_subscribers = subscribers['is_active'].sum()

active_rate = (active_subscribers / total_subscribers) * 100

print(f"Total Subscribers: {total_subscribers:,}")
print(f"Active Subscribers: {active_subscribers:,}")
print(f"Active Rate: {active_rate:.2f}%")
```

```python
#calculating KPI 3 — Churn Rate for that formula will be used (Inactive subscriber/Total subscriber)x100


inactive_subscribers = (~subscribers['is_active']).sum()

churn_rate = (inactive_subscribers / total_subscribers) * 100

print(f"Inactive Subscribers: {inactive_subscribers:,}")
print(f"Churn Rate: {churn_rate:.2f}%")
```

```python
#Calculating KPI 4 — Avg Completion Rate for that we need

#AVG(completion__pct)

#validating the columns


print(watch_history['completion_pct'].describe())
print(watch_history['completion_pct'].isna().sum())


```

```python
#calculating Avg Completion rate

avg_completion_rate = watch_history['completion_pct'].mean()

print(f"Avg Completion Rate: {avg_completion_rate:.2f}%")
```

```python
#calculating the KPI 5 — Monthly Recurring Revenue (MRR).

#validating the required columns

print(subscribers[['monthly_price_usd', 'is_active']].info())
print("\nMissing values:")
print(subscribers[['monthly_price_usd', 'is_active']].isna().sum())
```

```python
#calculating Monthly Recurring Revenue (MRR)

mrr = subscribers.loc[
    subscribers['is_active'],
    'monthly_price_usd'
].sum()

print(f"Monthly Recurring Revenue (MRR): ${mrr:,.2f}")
```

```python
#CALCULATING KPI 6 — ARPU (Average Revenue Per User).for that formulat
# ARPU = MRR / Active Subscribrs


arpu = mrr / active_subscribers

print(f"ARPU: ${arpu:.2f} per active subscriber")

```

```python
#calculating KPI 7 — Average Watch Time per Subscriber.for that formula 
#Avg Watch Time  = Total Watch Hours / Active Subscribers

avg_watch_time_per_subscriber = total_watch_hours / active_subscribers

print(f"Avg Watch Time per Active Subscriber: {avg_watch_time_per_subscriber:.2f} hours")
```

```python
#calculating KPI 8 — Watchlist Conversion

#for this we should identify
#watchlist entries
#watched entries



```

```python
#checking Watchlist Colums

print(watchlist.columns.tolist())
print("\nData types:")
print(watchlist.dtypes)
```

```python
print(watchlist['watched'].value_counts())
```

```python
#calculating watchlist conversion

total_watchlist_entries = len(watchlist)
watched_entries = watchlist['watched'].sum()

watchlist_conversion = (
    watched_entries / total_watchlist_entries) * 100

print(f"Watchlist Conversion: {watchlist_conversion:.2f}%")
```

```python
#calculating the KPI 9 — Hit Concentration.

#inspecting columns

print("Watch History columns:")
print(watch_history.columns.tolist())

print("\nTitles columns:")
print(titles.columns.tolist())
```

```python
# for Hit Concentration ,formula we need
#  (Plays from top 10% titles / Total plays) x 100

#checking  the play distribution

plays_per_title = watch_history['title_id'].value_counts()

total_unique_titles = plays_per_title.size

print(f"Total unique titles watched: {total_unique_titles}")
print(f"Total plays: {plays_per_title.sum():,}")
print(f"Top title plays: {plays_per_title.max():,}")
```

```python
# Top 10 titles

top_10_percent_count = int(total_unique_titles * 0.10)

top_10_titles = plays_per_title.head(top_10_percent_count)

top_10_plays = top_10_titles.sum()
total_plays = plays_per_title.sum()

print(f"Top 10% titles: {top_10_percent_count}")
print(f"Plays from top 10% titles: {top_10_plays:,}")
print(f"Total plays: {total_plays:,}")
```

```python
#Calculating Hit Conversion

hit_concentration = (
    top_10_plays / total_plays
) * 100

print(f"Hit Concentration: {hit_concentration:.2f}%")
```

```python
# Calculating KPI 10 — Originals Share of Hours.

# validating Is_original

print(titles['is_original'].value_counts())
print("\nMissing values:")
print(titles['is_original'].isna().sum())
```

```python
#we need to connect watch_history to titles using title_id

title_original_lookup = titles[['title_id', 'is_original']]

print(title_original_lookup.head())
print(f"\nRows in title lookup: {len(title_original_lookup):,}")
```

```python
#validating title_id

print(f"Total title rows: {len(title_original_lookup):,}")
print(f"Unique title IDs: {title_original_lookup['title_id'].nunique():,}")

print("Duplicate title IDs:",
    title_original_lookup['title_id'].duplicated().sum())
```

```python
#Joining the data

watch_with_original = watch_history.merge(
    title_original_lookup,
    on='title_id',
    how='left'
)

print(f"Watch history rows before join: {len(watch_history):,}")
print(f"Rows after join: {len(watch_with_original):,}")

print("\nMissing is_original after join:")
print(watch_with_original['is_original'].isna().sum())
```

```python
#alculating Original Content Watch Hours

original_watch_hours = (watch_with_original.loc[
        watch_with_original['is_original'],
        'watch_duration_min'].sum() / 60)

print(f"Original Content Watch Hours: {original_watch_hours:,.2f}")
```

```python
#Original Shares

#Originals Share = Original Watch Hours ÷ Total Watch Hours × 100

originals_share_hours = (
    original_watch_hours / total_watch_hours) * 100

print(f"Originals Share of Hours: {originals_share_hours:.2f}%")
```

```python
#calculating Cohort Retention.

#validating subscriber dates

print(subscribers[['signup_date', 'is_active']].info())

print("\nMissing values:")
print(subscribers[['signup_date', 'is_active']].isna().sum())
```

```python
#validating churn_date and the date range


print(subscribers[['signup_date', 'churn_date']].head())

print("\nChurn date data type:")
print(subscribers['churn_date'].dtype)

print("\nMissing churn dates:")
print(subscribers['churn_date'].isna().sum())

print("\nSignup date range:")
print(subscribers['signup_date'].min(), "to", subscribers['signup_date'].max())

print("\nChurn date range:")
print(subscribers['churn_date'].min(), "to", subscribers['churn_date'].max())
```

```python
subscribers['signup_date'] = pd.to_datetime(subscribers['signup_date'])
subscribers['churn_date'] = pd.to_datetime(subscribers['churn_date'])

print(subscribers[['signup_date', 'churn_date']].dtypes)
```

```python
#Creating the signup cohort month


subscribers['cohort_month'] = subscribers['signup_date'].dt.to_period('M')

print(subscribers[['signup_date', 'cohort_month']].head())
```

```python
#Creating 3-month checkpoint

subscribers['3_month_date'] = (subscribers['signup_date'] + pd.DateOffset(months=3))

print(subscribers[['signup_date', '3_month_date', 'churn_date']].head())
```

```python
#data cutoff date


data_end_date = max(subscribers['signup_date'].max(),subscribers['churn_date'].max())

print(f"Data End Date: {data_end_date.date()}")
```

```python
#3-month retained flag

subscribers['retained_3_months'] = (
    (subscribers['3_month_date'] <= data_end_date) &
    (subscribers['churn_date'].isna() |
        (subscribers['churn_date'] > subscribers['3_month_date'])))

print(subscribers['retained_3_months'].value_counts())
```

```python
#Separate eligible subscribers

eligible_3_months = subscribers['3_month_date'] <= data_end_date

subscribers['retained_3_months'] = (eligible_3_months &
    (subscribers['churn_date'].isna() |
        (subscribers['churn_date'] > subscribers['3_month_date'])))

print("Eligible for 3-month retention:", eligible_3_months.sum())
print("Retained at 3 months:", subscribers['retained_3_months'].sum())
```

```python
#3-month retention rate

retention_3_months = (
    subscribers['retained_3_months'].sum() /
    eligible_3_months.sum()) * 100

print(f"3-Month Retention Rate: {retention_3_months:.2f}%")
```

```python
#6-month checkpoint

subscribers['6_month_date'] = (
    subscribers['signup_date'] + pd.DateOffset(months=6))

print(subscribers[
        ['signup_date', '6_month_date', 'churn_date']].head())
```

```python
#6-month retained flag

eligible_6_months = subscribers['6_month_date'] <= data_end_date

subscribers['retained_6_months'] = (eligible_6_months & (subscribers['churn_date'].isna() |
        (subscribers['churn_date'] > subscribers['6_month_date'])))

print("Eligible for 6-month retention:", eligible_6_months.sum())
print("Retained at 6 months:", subscribers['retained_6_months'].sum())
```

```python
#6-month retention rate

retention_6_months = (subscribers['retained_6_months'].sum() /
    eligible_6_months.sum()) * 100

print(f"6-Month Retention Rate: {retention_6_months:.2f}%")
```

```python
#12-month checkpoint

subscribers['12_month_date'] = (subscribers['signup_date'] + pd.DateOffset(months=12))

print(subscribers[['signup_date', '12_month_date', 'churn_date']].head())
```

```python
#12-month retained flag


eligible_12_months = subscribers['12_month_date'] <= data_end_date

subscribers['retained_12_months'] = (eligible_12_months &
    (subscribers['churn_date'].isna() |
        (subscribers['churn_date'] > subscribers['12_month_date'])))

print("Eligible for 12-month retention:", eligible_12_months.sum())
print("Retained at 12 months:", subscribers['retained_12_months'].sum())
```

```python
#12-month retention rate

retention_12_months = (subscribers['retained_12_months'].sum() /
    eligible_12_months.sum()) * 100

print(f"12-Month Retention Rate: {retention_12_months:.2f}%")
```

```python
#Optional Cohort Retention table

cohort_retention_results = pd.DataFrame({
    'Retention Milestone': [
        '3-Month Retention',
        '6-Month Retention',
        '12-Month Retention'],
    'Eligible Subscribers': [
        eligible_3_months.sum(),
        eligible_6_months.sum(),
        eligible_12_months.sum()],
    'Retained Subscribers': [
        subscribers['retained_3_months'].sum(),
        subscribers['retained_6_months'].sum(),
        subscribers['retained_12_months'].sum()],
    'Retention Rate': [
        retention_3_months,
        retention_6_months,
        retention_12_months]})

cohort_retention_results['Retention Rate'] = (
    cohort_retention_results['Retention Rate'].map(
        lambda x: f"{x:.2f}%"))

cohort_retention_results.style.hide(axis="index")
```
